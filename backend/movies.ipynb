```python
features = [
    'title',
    'genres',
    'overview',
    'tagline',
    'original_language',
    'popularity',
    'vote_average'
]
```

```python
df = df.dropna(subset="title")
```

```python
df['overview'] = df["overview"].fillna('')
```

```python
df['tagline'] = df["tagline"].fillna('')
```

```python
df['original_language'] = df['original_language'].fillna('')
```

```python
df['genres'] = df['genres'].apply(lambda x: ', '.join([genre['name'] for genre in eval(x)]))
```

```python
df['tag'] = df['overview']+" " + df['genres']+" "+ df['tagline']+" "+df["original_language"]
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    words = text.split()
    words = [word for word in words if word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words]
    return " ".join(words)
```

```python
df['tag'] = df['tag'].apply(preprocess_text)
```

```python
df = df.reset_index(drop = True)
```

```python
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
```

```python
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(ngram_range =(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['tag'])
```

```python
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title,n=7):
  if title not in indices:
    return "Title not found"
  idx = indices[title]
  cos_sim = cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
  similar_idx = cos_sim.argsort()[::-1][1:n+1]
  return df["title"].iloc[similar_idx]
```

```python
recommend('Toy Story')
```

```python
import pickle
pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))
df.to_pickle('df.pkl')
pickle.dump(indices,open('indices.pkl','wb'))
pickle.dump(tfidf,open('tfidf.pkl','wb'))
```
